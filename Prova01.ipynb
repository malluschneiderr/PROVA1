#### Uma empresa deseja implementar um pequeno sistema para controlar os chamados de suporte recebidos ao longo do dia. Os chamados devem ser atendidos na ordem em que chegam. Entretanto, caso um atendimento tenha sido realizado por engano, o sistema deve permitir desfazer o último atendimento realizado, devolvendo esse chamado para o início da fila. 
#### Cada chamado possui: 
#### - codigo
#### - cliente 
#### - prioridade
#### considere inicialmente os seguintes chamados, nessa ordem de chegada: 
#### 101 - Ana - Média
#### 102 - Bruno - Alta
#### 103 - Carla - Baixa 
#### 104 - Daniel - Alta

## Parte 1 - Pensamento computacional

#### Qual estrutura de dados estudada é mais adequada para armazenar os chamados aguardando atendimento? Justifique.
#### **Resposta:** A estrutura de dados mais adequada para armazenar os chamados é a fila (Queue), pois os chamados precisam ser atendidos na mesma ordem em que chegam. Essa estrutura segue o princípio FIFO (First In, First Out), ou seja, o primeiro chamado que entra é o primeiro a ser atendido. Como o sistema também precisa permitir desfazer o último atendimento, podemos guardar o chamado atendido para que, caso seja necessário, ele volte para o início da fila, mantendo a ordem dos outros chamados.

#### Qual estrutura é a mais adequada para guardar os chamados já atendidos, considerando que o sistema precisa desfazer primeiro o atendimento mais recente? Justifique. 
#### **Resposta:** A estrutura de dados mais adequada é a pilha (Stack), pois ela segue o princípio LIFO (Last In, First Out), ou seja, o último elemento que entra é o primeiro a sair. Dessa forma, podemos guardar os chamados já atendidos e, caso seja necessário desfazer um atendimento, o sistema consegue recuperar primeiro o chamado atendido mais recentemente.

#### Sem executar o código indique como estarão as duas estruturas após a sequência a seguir: 
#### - Atender 
#### - Atender

#### - Atender 
#### **Resposta:** Fila de chamados: 103 – Carla – Baixa; 104 – Daniel – Alta.
#### Pilha de chamados atendidos: 101 – Ana – Média; 102 – Bruno – Alta (no topo).
#### Justificativa: Ana e Bruno são atendidos nessa ordem. Ao desfazer o último atendimento, Bruno retorna ao início da fila. Em seguida, Bruno é atendido novamente e volta para a pilha.

## Parte 2 - Implementação 
#### Implemente em Python uma classe chamada: **CentralAtendimento**
#### A classe deverá manter internamente: 
#### - uma estrutura para os chamados aguardando atendimento;
#### - uma estrutura para os chamados já atendidos.
#### Implemente obrigatóriamente os seguintes métodos: 

adicionar_chamado(codigo, cliente, prioridade 
atender_proximmo()
desfazer_ultimo_atendimento()
buscar_chamado(codigo)
exibir_situacao()

#### adicionar_chamado() - Adiciona o chamado ao final da fila.
#### atender_proximmo() - Remove o primeiro chamado da fila de espera e registra esse chamado entre os atendidos.
#### desfazer_ultimo_atendimento() - Remove o último chamado atendido e o recoloca no início da fila de espera. 
#### buscar_chamado(codigo) - Procura um chamado que ainda esteja aguardando atendimento e retorna seus dados. A busca deve ser implementada pelo próprio estudante, percorrendo os elementos da estrutura.
#### exibir_situacao() - Mostra os códigos dos chamados aguardando e atendidos.
#### Não é necessário criar uma classe específica para Chamado. Cada chamado pode ser representado por uma estrutura Python simples, como dicionário, tupla ou lista. 

```python
from collections import deque

class CentralAtendimento:
    def __init__(self):
        self.fila = deque()
        self.atendidos = []

    def adicionar_chamado(self, codigo, cliente, prioridade):
        chamado = {
            "codigo": codigo,
            "cliente": cliente,
            "prioridade": prioridade
        }
        self.fila.append(chamado)
        print("Chamado adicionado com sucesso!")

    def atender_proximo(self):
        if not self.fila:
            print("Não há chamados aguardando atendimento.")
            return

        chamado = self.fila.popleft()
        self.atendidos.append(chamado)
        print(f"Chamado {chamado['codigo']} atendido.")

    def desfazer_ultimo_atendimento(self):
        if not self.atendidos:
            print("Não há atendimentos para desfazer.")
            return

        chamado = self.atendidos.pop()
        self.fila.appendleft(chamado)
        print(f"Atendimento do chamado {chamado['codigo']} desfeito.")

    def buscar_chamado(self, codigo):
        for chamado in self.fila:
            if chamado["codigo"] == codigo:
                return chamado

        return None

    def exibir_situacao(self):
        print("Chamados aguardando atendimento:")
        for chamado in self.fila:
            print(chamado["codigo"])

        print("Chamados atendidos:")
        for chamado in self.atendidos:
            print(chamado["codigo"])

central = CentralAtendimento()

central.adicionar_chamado(101, "Ana", "Média")
central.adicionar_chamado(102, "Bruno", "Alta")
central.adicionar_chamado(103, "Carla", "Baixa")
central.adicionar_chamado(104, "Daniel", "Alta")

central.exibir_situacao()

central.atender_proximo()
central.atender_proximo()

central.exibir_situacao()

central.desfazer_ultimo_atendimento()

central.exibir_situacao()

resultado = central.buscar_chamado(102)

if resultado:
    print("Chamado encontrado:", resultado)
else:
    print("Chamado não encontrado na fila.")
```
